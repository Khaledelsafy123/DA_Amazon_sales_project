```python
# line plot (Year & total oredrs)

total_orders = (
    df.groupby("Year")["OrderID"]
      .nunique()
      .sort_index()
)

# Create Figure
plt.figure(figsize=(10,6))

# ----------------------------------------------------------
# Draw Line Plot
# ----------------------------------------------------------

plt.plot(

    total_orders.index,                 # X-axis values (Year)

    total_orders.values,                # Y-axis values (total oredrs)

    color="steelblue",              # Line color

    linewidth=2.5,                  # Line thickness

    linestyle="-",                  # Line style

    marker="o",                     # Marker shape

    markersize=6,                   # Marker size

    markerfacecolor="white",        # Marker fill color

    markeredgecolor="steelblue",    # Marker border color

    alpha=0.9,                      # Transparency


)

# ==========================================================
# Figure Customization
# ==========================================================

plt.title(
    "total oredrs by year",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "year",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "total orders",
    fontsize=13,
    fontweight="bold"
)

plt.xticks(range(2020, 2025,1), fontsize=11)
plt.yticks(range(19800, 20101,50),fontsize=11)

plt.grid(
    linestyle="--",
    alpha=0.4
)


plt.tight_layout()

plt.show()
```

```python
# Bar plot (Category & Total orders)

counts = df["Category"].value_counts()

# Create Figure
plt.figure(figsize=(10, 10))

# ----------------------------------------------------------
# Draw Bar Plot
# ----------------------------------------------------------

plt.bar(

    x=counts.index,              # Category labels
    height=counts.values,        # Frequency of each category
    width=0.6,                   # Width of each bar
    color=["steelblue", "salmon"],   # Fill color
    edgecolor="black",           # Border color
    linewidth=1.5,               # Border thickness
    alpha=0.8,                   # Transparency
    align="center"               # Center bars on x-ticks

)

# ==========================================================
# Figure Customization
# ==========================================================

# Chart Title
plt.title(
    "Total Orders by category",
    fontsize=18,
    fontweight="bold"
)

# Axis Labels
plt.xlabel(
    "category",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "total orders",
    fontsize=13,
    fontweight="bold"
)

# X-axis Ticks
plt.xticks(
    fontsize=11
)

# Y-axis Ticks
plt.yticks(
    range(0,int(counts.max()) + 1000,50)
    ,fontsize=11
)
plt.ylim(16000,int(counts.max())+50 )
# Grid
plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.4
)


# Automatically adjust spacing
plt.tight_layout()

# Display Figure
plt.show()
```

```python
# Scatter (Total orders & Total Quantity)

seller_data = df.groupby("SellerID").agg(
    TotalOrders=("OrderID", "nunique"),
    TotalQuantity=("Quantity", "sum")
).sort_values("TotalOrders", ascending=False)

plt.figure(figsize=(12, 8))

plt.scatter(
    x=seller_data["TotalOrders"],
    y=seller_data["TotalQuantity"],
    alpha=0.6
)

plt.title(
    "Seller Orders vs Total Quantity",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Total Orders",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Total Quantity",
    fontsize=13,
    fontweight="bold"
)

plt.grid(
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()
plt.show()
```

```python
# Scatter (Total orders & Total amount)

seller_data = df.groupby("SellerID").agg(
    TotalOrders=("OrderID", "nunique"),
    TotalAmount=("TotalAmount", "sum")
).sort_values("TotalOrders", ascending=False)

plt.figure(figsize=(12, 8))

plt.scatter(
    x=seller_data["TotalOrders"],
    y=seller_data["TotalAmount"],
    alpha=0.6
)

plt.title(
    "Seller Orders vs Total Amount",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Total Orders",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Total Amount",
    fontsize=13,
    fontweight="bold"
)

plt.grid(
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()
plt.show()
```

KEY BUSINESS INSIGHTS

__________________
Best

• Best Product: LED Desk Lamp is the most ordered product.

• Best Category: Electronics leads with 16,853 orders.

• Best Brand: CoreTech has the highest sales at around 8.57M.

• Best Market: The US has the highest number of orders (70,058).

• Best State: Texas leads with 24,896 orders.

________________

Worst

• Lowest Category: Clothing has the lowest order volume.

• Lowest Market: Australia has the lowest number of orders (4,130).

• Customer Retention: Some customers make multiple orders, so loyalty offers can help increase repeat purchases.

• Seller Performance: There are 1,999 sellers, so identifying low-performing sellers can help improve overall performance.

• Customer Behavior: Most customers made only 1 order, while a smaller number of customers made repeat purchases.
________________

Business Recommendation

Focus on Electronics, CoreTech, and the US market, especially Texas, while improving weaker markets and categories and encouraging repeat purchases.
